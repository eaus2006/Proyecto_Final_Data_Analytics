Promedio de valor de propiedad por ubicación:

```python
average_sale_price_bronx = database_bronx['SALE PRICE'].mean()
print(f"El promedio de 'SALE PRICE' en Bronx es: {average_sale_price_bronx:,.2f}")

average_sale_price_brooklyn = database_brooklyn['SALE PRICE'].mean()
print(f"El promedio de 'SALE PRICE' en Brooklyn es: {average_sale_price_brooklyn:,.2f}")

average_sale_price_manhattan = database_manhattan['SALE PRICE'].mean()
print(f"El promedio de 'SALE PRICE' en Manhattan es: {average_sale_price_manhattan:,.2f}")

average_sale_price_queens = database_queens['SALE PRICE'].mean()
print(f"El promedio de 'SALE PRICE' en Queens es: {average_sale_price_queens:,.2f}")

average_sale_price_statenisland = database_statenisland['SALE PRICE'].mean()
print(f"El promedio de 'SALE PRICE' en Staten Island es: {average_sale_price_statenisland:,.2f}")
```

```python
average_price_mapping = {
    'Bronx': average_sale_price_bronx,
    'Brooklyn': average_sale_price_brooklyn,
    'Manhattan': average_sale_price_manhattan,
    'Queens': average_sale_price_queens,
    'Staten Island': average_sale_price_statenisland
}
database['average_sale_price'] = database['neighbourhood_group'].map(average_price_mapping)
database['average_sale_price'] = database['average_sale_price'].round(2)
display(database.head())
```

Verificación de datos

```python
unique_neighborhood_prices = database[['neighbourhood_group', 'average_sale_price']].drop_duplicates()
display(unique_neighborhood_prices)
```

Crime Rate (siendo 5 el lugar con mayor tasa de criminalidad):

```python
crime_rate_mapping = {
    'Staten Island': 5,
    'Brooklyn': 4,
    'Queens': 3,
    'Manhattan': 2,
    'Bronx': 1
}
database['crime_rate'] = database['neighbourhood_group'].map(crime_rate_mapping)
display(database.head())
```

### EDA Inicial: Distribuciones

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Distribution of rental price
plt.figure(figsize=(10, 6))
sns.histplot(database['price'], bins=50, kde=True)
plt.title('Distribución del Precio de Alquiler')
plt.xlabel('Precio')
plt.ylabel('Frecuencia')
plt.xlim(0, 1500)
plt.show()

# Distribution of availability_365
plt.figure(figsize=(10, 6))
sns.histplot(database['availability_365'], bins=30, kde=True)
plt.title('Distribución de Disponibilidad (días por año)')
plt.xlabel('Disponibilidad (días)')
plt.ylabel('Frecuencia')
plt.show()
```

#### Interpretación: Distribución de Precios y Disponibilidad
- **Precio de Alquiler:** El gráfico muestra que la mayoría de los alquileres se concentran en rangos de precios bajos a moderados (por debajo de $500), con una cola larga hacia precios más altos, indicando menos propiedades de lujo.
- **Disponibilidad:** Se observa una distribución variada, con picos en propiedades que están disponibles casi todo el año (cerca de 365 días), lo que podría indicar alquileres a largo plazo o propiedades gestionadas profesionalmente, y también un número significativo de propiedades con baja disponibilidad.

### EDA Inicial: Correlaciones

```python
numerical_cols = database[['price', 'availability_365', 'average_sale_price', 'crime_rate']]

correlation_matrix = numerical_cols.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Matriz de Correlación de Características Numéricas')
plt.show()
```

#### Interpretación: Matriz de Correlación
- **Precio de Alquiler vs. Precio de Venta Promedio (0.18):** Hay una correlación positiva débil. Aunque no es fuerte, sugiere que las propiedades en áreas con precios de venta más altos tienden a tener alquileres ligeramente más elevados.
- **Precio de Alquiler vs. Tasa de Criminalidad (-0.12):** Una correlación negativa muy débil indica que, en general, un ligero aumento en la tasa de criminalidad (según nuestra escala) se asocia con una pequeña disminución en los precios de alquiler.
- **Precio de Venta Promedio vs. Tasa de Criminalidad (-0.79):** Existe una fuerte correlación negativa. Las zonas con propiedades más caras (mayor precio de venta) tienen significativamente menores tasas de criminalidad percibidas, lo cual es esperado.

### Visualización 1: Precio de venta promedio por precio de alquiler

```python
import matplotlib.ticker as mticker
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 7))
sns.scatterplot(x='average_sale_price', y='price', data=database, alpha=0.6)
plt.title('Precio de Alquiler vs. Precio de Venta Promedio de Propiedad')
plt.xlabel('Precio de Venta Promedio (millones de USD)')
plt.ylabel('Precio de Alquiler (USD)')

def millions_formatter(x, pos):
    return f'{x/1000000:.0f}M'

formatter = mticker.FuncFormatter(millions_formatter)
plt.gca().xaxis.set_major_formatter(formatter)

plt.show()
```

```python
highest_prices = database.nlargest(10, 'price')
display(highest_prices)
```

#### Interpretación: Precio de Alquiler vs. Precio de Venta Promedio
- El gráfico de dispersión confirma la débil correlación positiva observada en la matriz. Se puede apreciar que hay una tendencia general donde a mayor precio de venta promedio en un vecindario, el precio de alquiler tiende a ser un poco más alto, pero con mucha dispersión, indicando otros factores influyentes. Además, se ve que hay errores en la data en la parte del precio de los airbnb ya qye hay datos que superan los $8000 lo cual no es coherente.

### Visualización 2: Ubicación (neighbourhood_group) por precio de alquiler

```python
# Box plot of price by neighbourhood_group
plt.figure(figsize=(12, 7))
sns.boxplot(x='neighbourhood_group', y='price', data=database)
plt.title('Precio de Alquiler por Grupo de Vecindario')
plt.xlabel('Grupo de Vecindario')
plt.ylabel('Precio de Alquiler (USD)')
plt.ylim(0, 1000) # Limit y-axis for better visualization of common prices
plt.show()
```

#### Interpretación: Precio de Alquiler por Grupo de Vecindario
- Este boxplot muestra claramente que **Manhattan** tiene los precios de alquiler medianos más altos y una mayor variabilidad (rango intercuartílico más amplio), así como un mayor número de valores atípicos (outliers) de precios muy altos. **Staten Island** y el **Bronx** presentan los precios medianos más bajos, lo que refleja las diferencias de mercado entre los distritos de Nueva York.

### Visualización 3: Grado de criminalidad por precio de alquiler

```python
# Box plot of price by crime_rate
plt.figure(figsize=(12, 7))
sns.boxplot(x='crime_rate', y='price', data=database)
plt.title('Precio de Alquiler por Tasa de Criminalidad')
plt.xlabel('Tasa de Criminalidad (1: Baja, 5: Alta)')
plt.ylabel('Precio de Alquiler (USD)')
plt.ylim(0, 1000) # Limit y-axis for better visualization of common prices
plt.show()
```

#### Interpretación: Precio de Alquiler por Tasa de Criminalidad
- El boxplot sugiere una tendencia inversa: los grupos de vecindarios con tasas de criminalidad más bajas (1 y 2, es decir, Bronx y Manhattan según tu mapeo) tienden a tener precios de alquiler más altos. Por el contrario, los vecindarios con tasas de criminalidad más altas (4 y 5, Brooklyn y Staten Island) muestran precios de alquiler generalmente más bajos, aunque la escala de precios está limitada a 1000 USD.

### Visualización 4: Tipo de habitación por precio de alquiler

```python
# Box plot of price by room_type
plt.figure(figsize=(12, 7))
sns.boxplot(x='room_type', y='price', data=database)
plt.title('Precio de Alquiler por Tipo de Habitación')
plt.xlabel('Tipo de Habitación')
plt.ylabel('Precio de Alquiler (USD)')
plt.ylim(0, 1000) # Limit y-axis for better visualization of common prices
plt.show()
```

#### Interpretación: Precio de Alquiler por Tipo de Habitación
- Como era de esperar, los alquileres de **'Entire home/apt' (casa/apartamento entero)** presentan los precios medianos más altos y la mayor dispersión, incluyendo los alquileres más caros. Las **'Private room' (habitaciones privadas)** tienen precios significativamente más bajos, y las **'Shared room' (habitaciones compartidas)** son las más económicas, lo cual se alinea con la lógica del mercado de alojamientos.

### Visualización 5: Disponibilidad por precio de alquiler

```python
# Scatter plot of availability_365 vs price
plt.figure(figsize=(12, 7))
sns.scatterplot(x='availability_365', y='price', data=database, alpha=0.6)
plt.title('Precio de Alquiler vs. Disponibilidad (días por año)')
plt.xlabel('Disponibilidad (días)')
plt.ylabel('Precio de Alquiler (USD)')
plt.ylim(0, 1000) # Limit y-axis for better visualization of common prices
plt.show()
```

#### Interpretación: Precio de Alquiler vs. Disponibilidad
- El gráfico de dispersión no muestra una correlación lineal fuerte entre la disponibilidad y el precio. Hay propiedades con precios variados para todas las gamas de disponibilidad. Esto sugiere que la disponibilidad por sí sola no es un predictor principal del precio, y que otros factores son más influyentes.
